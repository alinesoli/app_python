```python
#Crie um programa que:
#Peça ao usuário 5 números
#Armazene esses números em uma lista
#Depois use um laço para:
#Mostrar todos os números
#Mostrar a soma total
#Mostrar quantos números são pares

lista = []
soma = 0
contador = 0

for i in range(5):
    num = int(input('Insira um número: '))
    lista.append(num)
    soma += num
    if num % 2 == 0:
        contador += 1

print('Os números da lista são:', lista)
print('A soma dos números listados são:', soma)
print('A quantidade de números pares é:', contador)
```

```python
#Crie um programa que:

#Peça números ao usuário até ele digitar 0 (zero para parar)
#não aceite números negativos (se digitar, peça novamente)
#Armazene os números válidos em uma lista
#No final, mostre:
#A lista completa
#A média dos números
#O maior número digitado

lista_lista = []
soma = 0
maior = 0

while True:
    numero = int(input("Insira um número (0 para parar): "))

    if numero == 0:
        break
    elif numero < 0:
        continue
    else:
        lista_lista.append(numero)
        soma += numero

        if numero > maior:
            maior = numero

# cálculo da média (evita erro se lista estiver vazia)
if len(lista_lista) > 0:
    media = soma / len(lista_lista)
else:
    media = 0

print("Lista:", lista_lista)
print("Média:", media)
print("Maior número:", maior)
```

```python
# Crie um programa que:
# - Peça ao usuário para digitar vários números inteiros (um por vez).
# - O programa deve parar quando o usuário digitar 0.
# - Ao final, mostre:
#   - Quantos números foram digitados (sem contar o 0)
#   - A soma de todos os números
#   - O maior número digitado
#   - O menor número digitado
#   - Quantos números são pares e quantos são ímpares

contagem = 0
soma = 0
maior = None
menor = None
par = 0
impar = 0

while True:
    num = int(input('Insira um número inteiro, ou 0 para parar: '))

    if num == 0:
        break

    contagem += 1
    soma += num

    if maior is None or num > maior:
        maior = num

    if menor is None or num < menor:
        menor = num

    if num % 2 == 0:
        par += 1
    else:
        impar += 1

print('Quantidade:', contagem)
print('Soma:', soma)
print('Maior:', maior)
print('Menor:', menor)
print('Pares:', par)
print('Ímpares:', impar)
```

```python
# Exercício: Sistema de Desconto

# Faça um programa que:
# 1. Peça o valor de uma compra
# 2. Pergunte se o cliente é VIP (digite 's' para sim ou 'n' para não)

# Regras:
# - Se o valor for acima de 100:
#     - VIP → 20% de desconto
#     - Não VIP → 10% de desconto
# - Se o valor for até 100:
#     - VIP → 10% de desconto
#     - Não VIP → sem desconto

# 3. Mostre o valor final da compra

# Desafio extra:
# - Se o usuário digitar algo diferente de 's' ou 'n', mostrar "Opção inválida"


compra = float(input('Insira o valor da compra: '))
vip = input("cliente é VIP digite 's' para sim ou 'n' para não ").lower()

if compra <= 0 or (vip != 's' and vip != 'n'):
    print('Insira um valor válido')
if compra >= 100 and vip == 's':
    print('Valor da compra com desconto cliente VIP, gasto acima de R$ 100,00: R$',compra*0.80)
elif compra >= 100 and vip == 'n':
    print('Valor da compra com desconto, gasto acima de R$ 100,00: R$',compra*0.90)
elif compra < 100 and vip == 's':
    print('Valor da compra com desconto cliente VIP, gasto abaixo de R$ 100,00: R$',compra*0.90)
elif compra >= 0.1 and compra < 100 and vip == 'n':
    print('Valor da compra: R$',compra)
```

```python
#Anna Júlia está criando um sistema para calcular o Índice de Massa Corporal (IMC) e
#fornecer recomendações básicas. O programa deve receber o peso e a altura de uma pessoa e exibir
#o valor do IMC, além de indicar se está abaixo do peso, com peso normal ou acima do peso. Crie um programa que
#receba o peso (em kg) e a altura (em metros) e calcule o IMC usando a fórmula: IMC = peso / (altura ** 2)
#Depois, exiba o valor do IMC e uma mensagem indicando se está abaixo do
#peso (IMC < 18.5), peso normal (18.5 <= IMC < 25) ou acima do peso (IMC >= 25).


peso = float(input('Insira seu peso atual, ex: 65.5KG : '))
altura = float (input('Insira a sua altura em metros ex: 1.62 '))
imc = peso / (altura ** 2)

if imc < 18.5:
    print('Abaixo do peso')
elif imc <= 25:
    print('Peso normal')
else:
    print('Acima do peso')

print('O IMC é', imc)
```
